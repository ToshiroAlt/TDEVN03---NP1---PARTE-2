```python
nome = input('qual o seu nome?: ')
nota1 = float(input('informe a primeira nota: '))
nota2 = float(input('informe a segunda nota: '))


class Pessoa:
    def __init__(self, nome, nota1, nota2):
        self.nome = nome
        self.nota1 = nota1
        self.nota2 = nota2

    def calcular_media(self):
        return (self.nota1 + self.nota2) / 2

    def verificar_aprovacao(self):
        media = self.calcular_media()
        if media >= 7:
            return f'Aluno: {self.nome}\nNota 1: {self.nota1}\nNota 2: {self.nota2}\nResultado: APROVADO!'
        else:
            return f'Me desculpe {self.nome}, você está REPROVADO!'

aluno = Pessoa(nome, nota1, nota2)

print("\n--- Resultado ---")
print("Nome:", aluno.nome)
print("Nota 1:", aluno.nota1)
print("Nota 2:", aluno.nota2)
print("Média:", aluno.calcular_media())
print("Situação Final:\n", aluno.verificar_aprovacao())
```
